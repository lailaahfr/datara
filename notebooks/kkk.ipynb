```python
import os

def ukuran_folder(path):
    total, rinci = 0, []
    for akar, _, files in os.walk(path):
        for f in files:
            p = os.path.join(akar, f)
            s = os.path.getsize(p)
            total += s
            rinci.append((s / 1024**2, p))
    return total / 1024**2, sorted(rinci, reverse=True)

total_mb, rinci = ukuran_folder('../data')
print(f'Total data: {total_mb:.1f} MB')
for mb, p in rinci[:15]:
    print(f'{mb:8.1f} MB  {p}')
```
